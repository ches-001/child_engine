```python
import importlib
from Chessnut import Game
from kaggle_environments import make
from kaggle_environments.envs.chess import chess
from os import path

def html_renderer():
    __file__ = "c:/Users/user/AppData/Local/Programs/Python/Python311/Lib/site-packages/kaggle_environments/envs/chess/chess"
    jspath = path.abspath(path.join(path.dirname(__file__), "chess.js"))
    with open(jspath, encoding="utf-8") as g:
        return g.read()
```

```python
def minimax_agent(obs, depth: int, v: str="v1") -> str:
    game = Game(obs.board)
    fen_str = game.get_fen()
    _, move = importlib.import_module(f"modules.{v}.chess_agent").minimax_agent(fen_str, depth, False)
    return move

def mcts_agent(obs, v: str="v1") -> str:
    game = Game(obs.board)
    fen_str = game.get_fen()
    _, move = importlib.import_module(f"modules.{v}.chess_agent").mcts_agent(fen_str, 2, 30, 50000)
    return move
```

```python
env = make("chess", debug=False)
env.html_renderer = html_renderer
result = env.run([lambda obs : minimax_agent(obs, 4, v="v1"), lambda obs : minimax_agent(obs, 4, v="v2")])
env.render(mode="ipython", width=800, height=800)
env.reset()
```

```python
import random
from matplotlib import pyplot as plt

signs = [1, -1]
series = []

def random_fibonacci(n):
    if n == 0: return 0
    elif n == 1: return 1
    s1 = random.choice(signs)
    s2 = random.choice(signs)
    val = (s1 * random_fibonacci(n-1)) + (s2 * random_fibonacci(n-2))
    series.append(val)
    return val

n = 100
random_fibonacci(n), series
```
