```python
from Chessnut import Game
from kaggle_environments import make
from kaggle_environments.envs.chess import chess
from modules import chess_agent
from main import ChessEngine
from os import path

def html_renderer():
    __file__ = "c:/Users/user/AppData/Local/Programs/Python/Python311/Lib/site-packages/kaggle_environments/envs/chess/chess"
    jspath = path.abspath(path.join(path.dirname(__file__), "chess.js"))
    with open(jspath, encoding="utf-8") as g:
        return g.read()
    
chess.OPENINGS
```

```python
def chess_agent(obs, engine: ChessEngine, depth: int=4) -> str:
    game = Game(obs.board)
    fen_str = game.get_fen()
    move, _ = engine.get_best_move(fen_str, depth=depth)
    return move

env = make("chess", debug=False, state=chess.OPENINGS[3])

env.html_renderer = html_renderer

engine_path = "engine/chess_engine"
engine_w = ChessEngine(engine_path, use_tt=True)
engine_b = ChessEngine(engine_path, use_tt=True)

result = env.run([lambda obs : chess_agent(obs, engine_w, 5), lambda obs : chess_agent(obs, engine_b, 5)])
env.render(mode="ipython", width=800, height=800)
env.reset()
engine_w.exit_process()
engine_b.exit_process()
```
